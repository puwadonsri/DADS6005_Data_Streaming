# Kafka Producer
**DADS6005 Quiz 2 — KsqlDB**

ส่งข้อความไปยัง Kafka topic `topic1`

```python
from kafka import KafkaProducer

producer = KafkaProducer(bootstrap_servers=['localhost:29092'])
topic = 'topic1'

for i in range(1, 4):
    msg = f'Hello-{i} Kafka...'
    producer.send(topic, msg.encode())
    print(f'Sent: {msg}')

producer.flush()
print('\n[Done] All messages sent')
```
