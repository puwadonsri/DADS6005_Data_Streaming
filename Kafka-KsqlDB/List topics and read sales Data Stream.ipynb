- List all Kafka topics
- Consume messages from Debezium CDC topic (`source.public.db_food_coded2_stream`)

```python
from kafka import KafkaConsumer

bootstrap_servers = ['localhost:29092']
consumer = KafkaConsumer(bootstrap_servers=bootstrap_servers)
topics = consumer.topics()
consumer.close()

print("Kafka Topics:")
for t in sorted(topics):
    print(f"  - {t}")
```

## Consume Debezium CDC Stream

อ่าน messages แบบ AVRO/JSON จาก topic ที่ Debezium สร้างไว้

```python
import json

topic = 'source.public.db_food_coded2_stream'
consumer = KafkaConsumer(
    topic,
    auto_offset_reset='earliest',
    bootstrap_servers=['localhost:29092']
)

count = 0
for msg in consumer:
    data = json.loads(msg.value.decode('utf-8'))
    after = data.get('payload', {}).get('after', {})
    op = data.get('payload', {}).get('op', '?')
    print(f"[op={op}] id={after.get('ids')} | reasons={str(after.get('comfort_food_reasons', ''))[:40]}")
    count += 1
    if count >= 5:
        print("\n[Sample] First 5 messages shown — stop here")
        break

consumer.close()
```
