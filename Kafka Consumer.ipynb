# Kafka Consumer
**DADS6005 Quiz 2 — KsqlDB**

อ่านข้อความจาก Kafka topic `topic1`

```python
from kafka import KafkaConsumer

consumer = KafkaConsumer(
    'topic1',
    auto_offset_reset='earliest',
    bootstrap_servers=['localhost:29092']
)

count = 0
for msg in consumer:
    print(f"Topic={msg.topic}, Message={msg.value}")
    count += 1
    if count >= 5:
        print("\n[Stopped] 5 messages consumed")
        break

consumer.close()
```
